## 1. 화면에 글자와 숫자 보여 주기

print는 괄호 안을 보여 줍니다. 글자는 큰따옴표로 감싸고 숫자는 그대로 씁니다. 괄호·따옴표는 영어 입력 상태에서 넣습니다. `#` 뒤의 설명은 실행되지 않습니다.

```python
print("안녕하세요")  # 따옴표 안의 글자를 보여 줍니다.
print(2 + 3)  # 더한 결과 5를 보여 줍니다.
```

**확인:** 첫 줄에는 안녕하세요, 둘째 줄에는 5가 나옵니다.

## 2. 이름 붙여 저장하기

=는 같다는 문제가 아니라 값을 저장하는 기호입니다. 숫자는 계산하고 글자는 따옴표로 구분합니다.

```python
temperature = 400  # 숫자 400에 온도라는 이름을 붙입니다.
next_temperature = temperature + 5  # 기존 숫자에 5를 더한 결과를 다른 이름에 담습니다.
print(next_temperature)  # 405가 나옵니다.
```

## 3. 여러 숫자 묶기

[ ]는 목록을 만듭니다. Python은 순서 번호를 0부터 셉니다. len은 묶음 길이를 셉니다.

```python
values = [10, 20, 30]  # 숫자 세 개를 묶습니다.
print(values[0])  # 0번, 즉 맨 첫 값 10을 봅니다.
print(values[1])  # 1번, 즉 두 번째 값 20을 봅니다.
print(len(values))  # 값이 세 개이므로 3을 보여 줍니다.
```

## 4. 같은 일 반복하기

for 아래의 들여쓴 줄을 값마다 실행합니다. 콜론 :과 공백 네 칸을 함께 봅니다. 반복 뒤의 들여쓰지 않은 print는 한 번만 실행됩니다.

```python
total = 0  # 합을 0에서 시작합니다.
for value in values:  # 10, 20, 30을 하나씩 value에 넣습니다.
    total = total + value  # 현재 숫자를 누적 합에 더합니다.
print(total)  # 반복이 끝난 뒤 합 60을 보여 줍니다.
```

## 5. 조건에 따라 다르게 하기

==는 같은지, >는 큰지 묻습니다. 조건 결과는 True 또는 False입니다.

```python
print(temperature == 400)  # 온도가 400과 같은지 물어 True가 나옵니다.
if temperature > 410:  # 410보다 높은 경우만 아래 줄을 실행합니다.
    print("연습 기준 초과")  # 이번에는 실행되지 않습니다.
else:  # 위 조건이 거짓이면 아래 줄을 실행합니다.
    print("연습 기준 이내")  # 이번에는 이 문장이 나옵니다.
```

## 6. 숫자 묶음을 한꺼번에 계산하기

목록에 2를 곱하면 반복 목록이 됩니다. 여기서 원하는 숫자별 곱셈은 NumPy 배열을 사용합니다. import ... as ...는 도구에 짧은 별명을 붙입니다.

```python
import numpy as np  # NumPy를 np라고 부릅니다.
numbers = np.array([1.0, 2.0, 3.0])  # 계산용 숫자 배열을 만듭니다.
print(numbers * 2)  # 각 숫자에 2를 곱해 [2, 4, 6]이 나옵니다.
print(numbers ** 2)  # 각 숫자를 제곱해 [1, 4, 9]가 나옵니다.
print(np.mean(numbers))  # 평균 2가 나옵니다.
```

## 7. 표에서 값 꺼내기

pandas의 DataFrame은 행과 열이 있는 표입니다. 점 뒤의 이름은 도구가 가진 기능입니다. head()처럼 괄호가 있으면 그 기능을 실행합니다. shape처럼 괄호가 없으면 저장된 정보를 읽습니다.

```python
import pandas as pd  # 표 도구를 pd라고 부릅니다.
table = pd.DataFrame({"temperature": [390, 400, 410], "thickness": [97, 100, 103]})  # 열 이름과 값 묶음으로 표를 만듭니다.
print(table.head())  # 표의 앞부분을 봅니다.
print(table["temperature"])  # 온도 열 하나를 꺼냅니다.
print(table[["temperature"]])  # 온도 열이 있는 작은 표로 꺼냅니다.
print(table.shape)  # (3, 2), 즉 세 행과 두 열입니다.
```

## 자주 보는 표기

| 표기 | 읽는 법 |
|---|---|
| `=` / `==` | 저장 / 같은지 비교 |
| `()` | 도구에 주는 값 또는 묶는 괄호 |
| `[]` | 목록 만들기 또는 일부 꺼내기 |
| `{이름: 값}` | 이름과 값을 짝지은 사전 |
| `.` | 도구나 표가 가진 기능 꺼내기 |
| `*`, `/`, `**` | 곱하기, 나누기, 거듭제곱 |
| `[:, 1]` | 모든 행의 두 번째 열 |
| `[:3]` | 앞에서 세 개 |
| `name=value` (함수 괄호 안) | 이름이 name인 설정에 value를 주기 |
| `True`, `False` | 맞음, 아님 |
| `_`가 들어간 이름 | 여러 단어를 연결한 이름 |
